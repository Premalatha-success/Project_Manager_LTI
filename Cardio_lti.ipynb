```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import gradio as gr

# Load dataset
url = "https://raw.githubusercontent.com/Premalatha-success/Datasets/refs/heads/main/CardioGoodFitness-1.csv"
df = pd.read_csv(url)

def generate_plot(plot_type, x_var, y_var, hue_var):
    fig, ax = plt.subplots(figsize=(8, 5))

    # Handle optional categorical hue
    hue = hue_var if hue_var != "None" else None

    try:
        if plot_type == "Histogram/Distribution":
            sns.histplot(data=df, x=x_var, hue=hue, kde=True, multiple="stack", ax=ax)
            ax.set_title(f"Distribution of {x_var}")

        elif plot_type == "Boxplot":
            sns.boxplot(data=df, x=x_var, y=y_var if y_var != "None" else None, hue=hue, ax=ax)
            ax.set_title(f"Boxplot of {x_var}" + (f" by {y_var}" if y_var != "None" else ""))

        elif plot_type == "Scatterplot":
            if y_var == "None":
                return "Scatterplot requires both X and Y variables. Please select a Y variable.", None
            sns.scatterplot(data=df, x=x_var, y=y_var, hue=hue, ax=ax)
            ax.set_title(f"Scatterplot: {x_var} vs {y_var}")

        elif plot_type == "Countplot":
            sns.countplot(data=df, x=x_var, hue=hue, ax=ax)
            ax.set_title(f"Count of {x_var}")

        plt.tight_layout()
        return "Plot rendered successfully.", fig
    except Exception as e:
        return f"Error rendering plot: {str(e)}", None

# List of columns for dropdowns
columns = list(df.columns)
numeric_columns = list(df.select_dtypes(include=['int64', 'float64']).columns)

# Gradio interface
with gr.Blocks() as demo:
    gr.Markdown("# CardioGoodFitness Data Visualization Dashboard")
    gr.Markdown("Explore the relationships and distributions within the CardioGoodFitness dataset.")

    with gr.Row():
        with gr.Column(scale=1):
            plot_type = gr.Dropdown(
                choices=["Histogram/Distribution", "Boxplot", "Scatterplot", "Countplot"],
                value="Histogram/Distribution",
                label="Select Plot Type"
            )
            x_var = gr.Dropdown(choices=columns, value="Age", label="X Axis Variable")
            y_var = gr.Dropdown(choices=["None"] + columns, value="None", label="Y Axis Variable (Optional/Required for Scatter)")
            hue_var = gr.Dropdown(choices=["None"] + ["Product", "Gender", "MaritalStatus"], value="None", label="Color By (Hue - Optional)")
            submit_btn = gr.Button("Generate Visualization")

        with gr.Column(scale=2):
            status_text = gr.Textbox(label="Status", interactive=False)
            plot_output = gr.Plot(label="Data Visualization")

    submit_btn.click(
        fn=generate_plot,
        inputs=[plot_type, x_var, y_var, hue_var],
        outputs=[status_text, plot_output]
    )

demo.launch(share=True)
```
